### `=== Partie 1 : Listes Python vs tableaux NumPy ===`

On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8

##### `1.1 Création d'une liste Python stockant ces températures`

```python
# liste standard de python
temperatures = [24.5, 25.1, 26.3, 27.0, 25.8]
print("liste stadard de python:")
print(temperatures)
```

##### `=== 1.2 Création d'un tableau NumPy équivalent stockant ces températures ===`

```python
import numpy as np

# tableau numpy
numpy_tableau = np.array(temperatures)
print("Tableau numpy")
print(numpy_tableau)
```

##### `=== 1.3 Multiplication par 2 sur ces deux collections === `

```python
import time

# Multiplication avec la liste pyrthon (boucle for)
start_time1 = time.time()
temperature_x2 = []
for temp in temperatures:
     temperature_x2.append(temp * 2)

print(f"Multiplication avec les listes python: \n {temperatures_x2} \n Temps d'execution: {time.time() - start_time1:.6f} \n") 

# Multiplication avec les tableaux numpy (operation vectoriel)
start_time2 = time.time()
print(f"Multiplication du tableau numpy: \n {numpy_tableau*2} \n Temps d'execution {time.time() - start_time2:.6f}")
```

#### `=== 1.4 Observation ===`

On constate qu'avec le calcul sur les listes standards de python, pour ce cas, que la durée d'execution est plus courte qu'avec le tableau `NumPy` mais en terme ligne de code, la multiplication avec le tableau numpy est beaucoup plus optimisé (1 ligne)

##### `8.2 Standardisation du tableau des températures`

```python
moyennes = np.mean(mes_24_heures)
ecart_type = np.std(mes_24_heures)

mes_24_heures_standar = (mes_24_heures - moyennes) / (ecart_type )
print(mes_24_heures_standar)
print(f"Moyenne après standardisation: \n  {np.round(np.mean(mes_24_heures_standar))}")
print(f"Ecart-type après standardisation: \n  {np.round(np.std(mes_24_heures_standar))}")
```

##### `8.3 Normalisation du tableau des températures`

```python
x_min = np.min(mes_24_heures)
x_max = np.max(mes_24_heures)

mes_normaliser = (mes_24_heures - x_min) / (x_max - x_min)

print(mes_normaliser)
```
